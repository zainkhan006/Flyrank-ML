## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

This slice has 30,000 pages across 32 clients, so there is enough inventory to study associations, not just a handful of anecdotes. About 55.99% of pages got atleast one click in the last 90 days, so being shown(aka impressions) and being clicked are not the same outcome, which is why my lane looks at both. Keyword "search_volume" has a correlation of 0.001 with impressions_90d, which reflects the near zero score in the starter notebooks, and this is a column people treat as the demand signal that barely travels with visibility here. That is why 7 weeks on which numbers actually move together is worth it, and why a one liner rule is not enough.

```python
import os, sys, subprocess
import pandas as pd

inColab = "google.colab" in sys.modules
repoURL = "https://github.com/zainkhan006/Flyrank-ML"
repoDirectory = "Flyrank-ML"

if inColab:
  if not os.path.isdir("data/raw"):
    if not os.path.isdir(repoDirectory):
      subprocess.run(["git", "clone", "--depth", "1", repoURL, repoDirectory], check=True)
    os.chdir(repoDirectory)

pages = pd.read_csv("data/raw/content_refresh_anonymized.csv")
noofpages = len(pages)
noofclients = pages["client_id"].nunique()
noofclicks = (pages["clicks_90d"] > 0).sum()
clickpercent = 100 * noofclicks / noofpages
volumeCorrelation = pages["search_volume"].corr(pages["impressions_90d"])
print("pages:", noofpages)
print("clients:", noofclients)
print("no of pages with atleast 1 click in the last 90 days:", round(clickpercent, 2), "%")
print("corrleation of keyword search volume with impressions_90d:", round(volumeCorrelation, 3))
```
